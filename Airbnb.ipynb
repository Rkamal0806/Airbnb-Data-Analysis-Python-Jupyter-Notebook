# Visualisations and Analysis

```python
# Idenfying outliers in price
df1 = df[df['price'] < 1500]

sns.boxplot(data=df1, x='price', color='green')
plt.show()
```

```python

#There are many outliers in price column where price goes beyond 1500 dollars
```

```python
df.info()
```

```python
#Price distribution
```

```python
plt.figure(figsize=(8,6))
sns.histplot(data=df1,x='price',bins=100)
plt.title('price distribution')
plt.ylabel("frequency")
plt.show()
```

```python
# Conclusion:max frequency lies between 0 to 500 dollars
```

```python
#Price distribuion
plt.figure(figsize=(6, 3))
sns.histplot(data=df, x='availability_365', color='red')
plt.title('availability_365 Distribuition')
plt.ylabel("Frequency")
plt.show()
```

```python
df.columns
```

```python
df.groupby(by='neighbourhood_group')['price'].mean()
```

```python
#Conclusion:Mean price of airbnb at a location like Bronx is 118.40 dollars. 
```

```python
df['price per bed']= df['price']/df['beds']
```

```python
df.groupby(by='neighbourhood_group')['price per bed'].mean()
```

```python
#Conclusion:price per bed of airbnb at a location like Bronx is 85.16 dollars.
```

```python
# Price dependency on neighbourhood
sns.barplot(data=df, x='neighbourhood_group', y='price', hue='room_type')
```

```python
#Conclusion:Manhattan has highest rates among all groups.& in Manhattan highest price goes for hotel room types, then comes shared rooms.
```

```python
#Which locality got how m any reviews - data is without outlier
#Number of reviews and price relations
plt.figure(figsize=(8, 5))
plt.title("Locality and Review Dependency")
sns.scatterplot(data=df1, x='number_of_reviews', y='price', hue='neighbourhood_group')
plt.show()
```

```python
#Properties of broklyn & queens got less price more number of reviews.Manhattan properties got more price less number of reviews
```

```python
sns.pairplot(data=df1, vars=['price', 'minimum_nights', 'number_of_reviews', 'availability_365'], hue='room_type')
```

```python
#Geographical Distribution of AirBnb Listing
plt.figure(figsize=(10, 7))
sns.scatterplot(data=df1, x='longitude', y='latitude', hue='room_type')
plt.title("Geographical Distribution of AirBnb Listing")
plt.show()
```

```python
#Heat map - correlation of one variable with others for numerical column

corr = df1[['latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month', 'availability_365', 'beds']].corr()
corr

plt.figure(figsize=(8, 6))
sns.heatmap(data=corr, annot=True)
```
